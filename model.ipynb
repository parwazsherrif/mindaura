PIPELINE AND COLUMN TRANSFER

```python
from sklearn.model_selection import train_test_split
skewed_col=['Study_Hours']                                    #logtransformation
other_numericcol=['Age', 'Avg_Daily_Usage_Hours', 'Daily_Unlocks', 'Physical_Activity_Hours', 'Sleep_Hours_Per_Night'] #standard scaler
ordinal_col=['Stress_Level'] #ordinal encoding 
normal_col=[ 'Gender',  'Academic_Level', 'Most_Used_Platform', 'Purpose_Of_Use', 'Grouped_Countries'] # one hot encoding
feature_col=skewed_col+other_numericcol+ordinal_col+normal_col
X=df[feature_col]
y=df['Mental_Health_Score']
X_train, X_test, y_train, y_test = train_test_split( X, y, test_size=0.30, random_state=42)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer,StandardScaler,LabelEncoder,OrdinalEncoder,OneHotEncoder
skew_pipeline=Pipeline(steps=[('log_transform',FunctionTransformer(func=np.log1p)),
('scale',StandardScaler())])
plain_col_pipeline=Pipeline(steps=[('scale',StandardScaler())])
oridinal_pipeline=Pipeline(steps=[('encode',OrdinalEncoder(categories=[['Low','Medium','High','Very High']]))])
nominal_pipeline=Pipeline(steps=[('one_hot',OneHotEncoder(handle_unknown='ignore'))])
```

BUILD A PREPROCESSOR

```python
from sklearn.compose import ColumnTransformer
preprocessor=ColumnTransformer(transformers=[
    ('Skewed_pipe',skew_pipeline,skewed_col),
    ('Plain_pipe',plain_col_pipeline,other_numericcol),
    ('Ordinal_pipe',oridinal_pipeline,ordinal_col),
    ('Nominal_pipe',nominal_pipeline,normal_col)
])
```

BASELINE PIPELINE 

```python
from sklearn.linear_model import LinearRegression
lr_pipeline=Pipeline(steps=[('preprocess',preprocessor),
('regressor',LinearRegression())])
```

```python
lr_pipeline.fit(X_train,y_train)
```

```python
lr_predict=lr_pipeline.predict(X_test)
lr_predict_train=lr_pipeline.predict(X_train)
```

EVALUTION OF MODEL

```python
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
lr_r2_testing=r2_score(y_test,lr_predict)
lr_r2_train=r2_score(y_train,lr_predict_train)
lr_mse=mean_squared_error(y_test,lr_predict)
print(f'ACCURACY OF TRAINING DATA: {lr_r2_train}')
print(f'ACCURACY OF TESTING DATA: {lr_r2_testing}')
print(f'MEAN ABSOLUTE ERROR: {lr_mse}')

```

RANDOM FOREST

```python
from sklearn.ensemble import RandomForestRegressor
rf_pipeline=Pipeline(steps=[('preprocess',preprocessor),
('random forest',RandomForestRegressor(random_state=42))])
rf_pipeline.fit(X_train,y_train)
```

```python
rf_predict_test=rf_pipeline.predict(X_test)
rf_predict_train=rf_pipeline.predict(X_train)
```

EVALUTION OF MODEL

```python
rf_test_r2=r2_score(y_test,rf_predict_test)
rf_train_r2=r2_score(y_train,rf_predict_train)
rf_mse=mean_squared_error(y_test,rf_predict_test)
print(f'ACCURACY OF TRAINING DATA: {rf_test_r2}')
print(f'ACCURACY OF TESTING DATA: {rf_train_r2}')
print(f'MEAN ABSOLUTE ERROR: {rf_mse}')
```

HYPER PARAMETER TUNING

```python
from sklearn.model_selection import RandomizedSearchCV
param_grid={
    'random forest__n_estimators':[100,200,300],
    'random forest__max_depth':[5,10,15],
    'random forest__min_samples_split':[2,5,10],
    'random forest__min_samples_leaf':[1,2,4]
}
tuning=RandomizedSearchCV(estimator=rf_pipeline,param_distributions=param_grid,n_iter=15,cv=5,scoring='r2',random_state=42,n_jobs=-1)
tuning
```

```python
tuning.fit(X_train,y_train)
```

```python
tuning.best_params_
```

```python
rf_best_pipeline=tuning.best_estimator_
rf_best_predict=rf_best_pipeline.predict(X_test)
rf_best_score=r2_score(y_test,rf_best_predict)
print(f'R2 Score for random forest after tuning is {rf_best_score}')
print(f'mean absolute error for random forest after tuning is {mean_absolute_error(y_test,rf_best_predict)}')
```

OVERALL EVALUTION OF MODEL

```python
from sklearn.metrics import mean_squared_error
lr_rmse=np.sqrt(mean_squared_error(y_test,lr_predict))
rf_rmse=np.sqrt(mean_squared_error(y_test,rf_predict_test))
rf_tuned_predicts=tuning.best_estimator_.predict(X_test)
rf_tuned_rmse=np.sqrt(mean_squared_error(y_test,rf_best_predict))
rf_tuned_train_predict=tuning.best_estimator_.predict(X_train)
rf_tuned_training_r2=r2_score(y_train,rf_tuned_train_predict)
rf_tuned_mse=mean_squared_error(y_test,rf_tuned_predicts)
rf_tuned_r2=r2_score(y_test,rf_tuned_predicts)
results=pd.DataFrame({
    "Model":['Linear Regression','Random Forest(default)','Random Forest(tuned)'],
    'Training R2':[lr_r2_train,rf_train_r2,rf_tuned_training_r2],
    'R2':[lr_r2_testing,rf_test_r2,rf_tuned_r2],
    "MAE":[lr_mse,rf_mse,rf_tuned_mse],
    'RMSE':[lr_rmse,rf_rmse,rf_tuned_rmse]
})
print(results)
```
